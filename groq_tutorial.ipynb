```python
from groq import Groq
```

```python
import os
from getpass import getpass

# Recommended: set GROQ_API_KEY as an environment variable.
# If it is not set, enter your API key when prompted.
GROQ_API_KEY = os.getenv("GROQ_API_KEY") or getpass("Enter your Groq API key: ")
```

```python
client = Groq(api_key=GROQ_API_KEY)
```

```python
chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "Tell me about Narendra",
        }
    ],
    model="openai/gpt-oss-120b",
)

print(chat_completion.choices[0].message.content)
```

```python
def ask_llm(user_query):
    message = [
        {
            "role": "user",
            "content": user_query
        }
    ]

    chat_completion = client.chat.completions.create(
        messages=message,
        model="openai/gpt-oss-20b"
    )

    response = chat_completion.choices[0].message.content
    return response
```

```python
ask_llm("tell me about CJP")
```

```python
history = []

while True:
    user_query = input("Enter user query: ")

    if user_query.lower() == "exit":
        break

    message = {
        "role": "user",
        "content": user_query
    }

    history.append(message)
    

    chat_completion = client.chat.completions.create(
        messages=history,
        model="openai/gpt-oss-20b"
    )

    response = chat_completion.choices[0].message.content

    print(response)

    ai_response = {
        "role": "assistant",
        "content": response
    }

    history.append(ai_response)
```

```python
ask_llm("What am I studying?")
```

```python
while True:
    user_query = input("Enter your interests: ")

    if user_query.lower() == "exit":
        print("Goodbye!")
        break

    prompt = f"""
You are a college club recommendation chatbot.

The student has the following interests:
{user_query}

Recommend ONE suitable college club by name.

Give the answer in exactly this format:
Club: <club name>
Reason: <one short reason>
"""

    chat_completion = client.chat.completions.create(
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        model="openai/gpt-oss-20b"
    )

    response = chat_completion.choices[0].message.content

    print(response)
```

       ↓
     Groq
